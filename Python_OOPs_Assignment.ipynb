**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, initial_balance=0):
     
        self.__balance = initial_balance


    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited: {amount}")
        else:
            print("Deposit amount zero ya negative nahi ho sakti.")

   
    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Successfully withdrew: {amount}")
            else:
                print("Error: Insufficient balance! Balance zero ya negative nahi ho sakta.")
        else:
            print("Withdraw amount positive honi chahiye.")

    
    def get_balance(self):
        return self.__balance


account = BankAccount(5000)


account.deposit(2000)   


account.withdraw(3000) 


account.withdraw(5000)  

# Final balance
print("Current Balance:", account.get_balance())
```

Jab aap kisi private variable (jiske shuru mein __ ho) ko class ke bahar se direct access karne ki koshish karte hain, toh Python AttributeError throw karta hai. Iska matlab hota hai ke object ke paas us naam ka koi public attribute mojood nahi hai.

```python
class Person:
    def __init__(self, name, age):
        self.name = name          # Public variable
        self.__age = age          # Private variable 

p1 = Person("Ahmed", 30)
print("Name (Public):", p1.name)

# 2. Private variable ko direct access karne ki koshish (Yahan Error aayega)
try:
    print("Age (Private):", p1.__age)
except AttributeError as e:
    print("\nError Caught:", e)
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
#Python mein super() function ka istemal child class ke andar se parent class ke constructor (__init__) ya methods ko call karne ke liye kiya jata hai. Iska sab se bara faida yeh hai ke aapko parent class ka naam baar-baar likhna nahi parta aur code maintain karna asaan ho jata hai.
# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child Class inheriting from Person
class Employee(Person):
    def __init__(self, name, age, emp_id, salary):
        # super() ke zariye parent class (Person) ke __init__ ko call kiya
        super().__init__(name, age)
        
        # Child class ke apne attributes
        self.emp_id = emp_id
        self.salary = salary

emp1 = Employee("Ali", 28, "EMP101", 75000)

# Dono classes ke attributes ko print karna
print("Name:", emp1.name)       # Parent class se mila
print("Age:", emp1.age)         # Parent class se mila
print("Employee ID:", emp1.emp_id) # Child class ka apna
print("Salary:", emp1.salary)   # Child class ka apna
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
#Hierarchical Inheritance OOP (Object-Oriented Programming) ki aik aisi qisam hai jismein aik hi parent class se aik se zyada child classes inherit karti hain.
import math

# Parent Class
class Shape:
    def __init__(self, color):
        self.color = color

    def display_color(self):
        return f"Shape ka color yeh hai: {self.color}"


# Child Class 1 (Circle inherits from Shape)
class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)  # Parent ka constructor call kiya
        self.radius = radius

    def calculate_area(self):
        return math.pi * (self.radius ** 2)


# Child Class 2 (Square inherits from Shape)
class Square(Shape):
    def __init__(self, color, side):
        super().__init__(color)  # Parent ka constructor call kiya
        self.side = side

    def calculate_area(self):
        return self.side ** 2

my_circle = Circle("Red", 5)
print("--- Circle ---")
print(my_circle.display_color())  # Parent class ka method
print(f"Radius: {my_circle.radius}, Area: {my_circle.calculate_area():.2f}")

print("\n--- Square ---")
my_square = Square("Blue", 4)
print(my_square.display_color())  # Parent class ka method
print(f"Side: {my_square.side}, Area: {my_square.calculate_area()}")
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Method Overriding Object-Oriented Programming (OOP) ka aik aisa feature hai jismein child class apne parent class ke kisi method ko apni zaroorat ya khasiyat ke mutabiq dubara define (re-write) karti hai.

# Parent Class
class Bird:
    def fly(self):
        print("Most birds can fly")

# Child Class inheriting from Bird
class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly")

b = Bird()
print("Normal Bird:")
b.fly() 


o = Ostrich()
print("\nOstrich:")
o.fly()  
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
#Operator Overloading ka matlab yeh hai ke hum built-in operators (jaise +, -, *, ==) ko apne custom classes ke objects par bhi kaam karne ke qabil banayein. Normal taur par operators integers, floats ya strings par kaam karte hain, lekin magic methods (jaise __add__) ke zariye hum yeh define kar sakte hain ke jab do custom objects ko + operator se add kiya jaye, toh kya hona chahiye
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # __add__ magic method operator overloading ke liye use hota hai (+)
    def __add__(self, other):
        # Yeh method tab chalega jab Point objects ke beech '+' lagaya jayega
        return Point(self.x + other.x, self.y + other.y)

    # Object ko print karne ke liye __str__ method
    def __str__(self):
        return f"Point({self.x}, {self.y})"


p1 = Point(2, 3)
p2 = Point(4, 5)


p3 = p1 + p2 

print(p3) 
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data using a portable battery and Core i7 processor! ")

class Desktop(Computer):
    def process(self):
        print("Desktop is processing data using a heavy-duty power supply and dedicated GPU! ")


my_laptop = Laptop()
my_desktop = Desktop()

my_laptop.process() 
my_desktop.process()  
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod

class Car(ABC):
    
    def start_engine(self):
        print("Car engine started. Vroom Vroom! ")

    @abstractmethod
    def drive(self):
        pass

class SportsCar(Car):

    def drive(self):
        print("Sports car is driving at super fast speed on the highway! ")

my_car = SportsCar()

my_car.start_engine()  

my_car.drive()         
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    
    def validate_amount(self, amount):
        if amount > 0:
            print(f"Amount Rs. {amount} is valid.")
            return True
        else:
            print("Invalid payment amount!")
            return False

    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def __init__(self, card_number, cvv):
        self.card_number = card_number
        self.cvv = cvv

    def process_payment(self, amount):
        if self.validate_amount(amount):
            # Masking card number for security
            masked_card = "****-****-****-" + self.card_number[-4:]
            print(f"Processing Rs. {amount} via Credit Card ({masked_card}). Payment Successful! ")


class UPIPayment(Payment):
    def __init__(self, upi_id):
        self.upi_id = upi_id

    def process_payment(self, amount):
        if self.validate_amount(amount):
            print(f"Processing Rs. {amount} via UPI ID ({self.upi_id}). Payment Successful! ")



def checkout(payment_method_obj, amount):
    print("\n--- Starting Checkout ---")
    payment_method_obj.process_payment(amount)


cc_payment = CreditCardPayment("4532123456789012", "123")
upi_payment = UPIPayment("user@ybl")

checkout(cc_payment, 5000)
checkout(upi_payment, 1500)
```
